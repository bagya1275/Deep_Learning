```python

import numpy as np
import matplotlib.pyplot as plt

# 1. Create Customer-Genre and Genre-Book matrices

customer_genre = np.array([
    [5, 3, 4, 2, 1],
    [4, 5, 3, 1, 2],
    [1, 2, 5, 4, 3],
    [2, 1, 4, 5, 3],
    [5, 4, 2, 3, 1],
    [3, 2, 4, 5, 1]
])

genre_book = np.array([
    [0.8, 0.2, 0.5, 0.3],
    [0.4, 0.9, 0.3, 0.6],
    [0.7, 0.5, 0.8, 0.2],
    [0.3, 0.6, 0.4, 0.9],
    [0.5, 0.3, 0.7, 0.8]
])

print("Customer-Genre Matrix (6x5):")
print(customer_genre)

print("\nGenre-Book Matrix (5x4):")
print(genre_book)


# 2. Dot product and Matrix multiplication

dot_product = np.dot(customer_genre[0], genre_book[:, 0])
score_matrix = np.matmul(customer_genre, genre_book)

print("\nDot Product:", dot_product)
print("\nMatrix Multiplication (6x4):")
print(score_matrix)


# Transpose
print("\nTranspose of Customer-Genre Matrix:")
print(customer_genre.T)

# Inverse (using a square matrix)
square_matrix = np.matmul(customer_genre.T, customer_genre)

inverse_matrix = np.linalg.inv(square_matrix)

print("\nInverse Matrix:")
print(inverse_matrix)

# Element-wise operations
print("\nElement-wise Addition:")
print(customer_genre + 2)

print("\nElement-wise Multiplication:")
print(customer_genre * customer_genre)


# 3. Cosine Similarity between customers

from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(customer_genre)

print("\nCosine Similarity Matrix:")
print(similarity)

# Find two most similar customers
np.fill_diagonal(similarity, -1)

i, j = np.unravel_index(
    np.argmax(similarity), similarity.shape
)

print("\nMost Similar Customers:")
print("Customer", i + 1, "and Customer", j + 1)
print("Similarity Score:", similarity[i, j])


# 4. Dense Layer: y = Wx + b

W = genre_book.T
x = customer_genre[0]
b = np.array([1, 1, 1, 1])

y = np.dot(W, x) + b

print("\nDense Layer Output (y = Wx + b):")
print(y)


# 5. Visualise Score Matrix as Heatmap

plt.figure(figsize=(8, 5))

plt.imshow(score_matrix, cmap="viridis", aspect="auto")

plt.colorbar(label="Recommendation Score")

plt.xlabel("Books")
plt.ylabel("Customers")
plt.title("Customer-Book Recommendation Score Heatmap")

plt.xticks(range(4), ["Book 1", "Book 2", "Book 3", "Book 4"])
plt.yticks(range(6), ["C1", "C2", "C3", "C4", "C5", "C6"])

plt.show()
```
