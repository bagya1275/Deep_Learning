```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import entropy

# Observed weather distribution over 30 days
weather = np.array([
    15, 8, 7
]) / 30

# Two prediction models
model1 = np.array([0.50, 0.30, 0.20])
model2 = np.array([0.30, 0.40, 0.30])

# Entropy
H = entropy(weather, base=2)

# Cross entropy
CE1 = -np.sum(weather * np.log2(model1))
CE2 = -np.sum(weather * np.log2(model2))

# KL divergence
KL1 = entropy(weather, model1, base=2)
KL2 = entropy(weather, model2, base=2)

print("Observed Entropy:", H)
print("\nModel 1 Cross Entropy:", CE1)
print("Model 1 KL Divergence:", KL1)

print("\nModel 2 Cross Entropy:", CE2)
print("Model 2 KL Divergence:", KL2)

print("\nVerification CE = H + KL")
print("Model 1:", CE1, "=", H + KL1)
print("Model 2:", CE2, "=", H + KL2)

# KL divergence is asymmetric
print("\nKL(P || Q):", entropy(weather, model1, base=2))
print("KL(Q || P):", entropy(model1, weather, base=2))

# Binary cross entropy
actual = 1
predicted = np.linspace(0.01, 0.99, 100)

loss = -(
    actual * np.log(predicted) +
    (1 - actual) * np.log(1 - predicted)
)

plt.figure(figsize=(8, 5))
plt.plot(predicted, loss)
plt.xlabel("Predicted Probability")
plt.ylabel("Binary Cross Entropy Loss")
plt.title("Binary Cross Entropy Loss")
plt.grid()
plt.show()
```
